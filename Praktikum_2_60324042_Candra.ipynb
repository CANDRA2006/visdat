**Operator Aritmatika**

```python
nilai_tugas = 85
nilai_uts = 78
nilai_uas = 90

# Nilai akhir = 30% tugas + 35% UTS + 35% UAS

nilai_akhir = (nilai_tugas * 0.3) + \
(nilai_uts * 0.35) + \
(nilai_uas * 0.35)
print("Nilai akhir:", nilai_akhir)
```

**Percabangan (if / elif / else)**

```python
if nilai_akhir >= 85:
  grade = "A"
elif nilai_akhir >= 70:
  grade = "B"
elif nilai_akhir >= 60:
  grade = "C"
else:
  grade = "D"

print("Grade Anda:", grade)
```

**tePerulangan for**

```python
nilai_kelas = [80, 75, 90, 65, 88]

for n in nilai_kelas:
  print("Nilai:", n)

# Menghitung total dan rata-rata
total = 0
for n in nilai_kelas:
  total += n

print("Total :", total)
print("Rata2 :", total / len(nilai_kelas))
```

**Perulangan while**

```python
i = 1

while i <= 5:
  print("Pertemuan ke-", i)
  i += 5 # jangan lupa menaikkan nilai i
```

**Membuat Fungsi dengan def**

```python
def hitung_rata2(data):
    """Menghitung nilai rata-rata dari sebuah list"""
    total = sum(data)
    jumlah = len(data)
    return total / jumlah

  # Memanggil Fungsi
nilai_kelas = [80, 75, 90, 65, 88]
hasil = hitung_rata2(nilai_kelas)

print("Rata-rata kelas:", hasil)
```

**Menggabungkan Fungsi dan Percabangan**

```python
def kategori_nilai(n):
  if n >= 85:
      return "Sangat Baik"
  elif n >= 70:
      return "Baik"
  else:
      return "Perlu Belajar Lagi"

# Terapkan ke seluruh nilai kelas
for n in nilai_kelas:
  print(n, "->", kategori_nilai(n))
```

```python
# Latihan 2
def nilai_tertinggi(data):
    tertinggi = data[0]

    for nilai in data:
        if nilai > tertinggi:
            tertinggi = nilai

    return tertinggi


# Penggunaan
data = [76, 85, 90, 100, 95, 88]

hasil = nilai_tertinggi(data)
print("Nilai tertinggi:", hasil)
```

```python
# Latihan 3
nilai_kelas = [90, 85, 78, 70, 65, 59, 45]

# Fungsi untuk menentukan kategori nilai
def kategori_nilai(nilai):
    if nilai >= 85:
        return "Sangat Baik"
    elif nilai >= 70:
        return "Baik"
    elif nilai >= 60:
        return "Cukup"
    else:
        return "Kurang"

# Terapkan ke seluruh nilai kelas
for n in nilai_kelas:
    print(n, "->", kategori_nilai(n))
```

```python
# Latihan 4
nilai_ujian = [89, 72, 65, 90, 95, 78, 60, 88]

# Variabel untuk menghitung jumlah mahasiswa
lulus = 0
tidak_lulus = 0

# Perulangan dan percabangan
for nilai in nilai_ujian:
    if nilai >= 70:
        lulus += 1
    else:
        tidak_lulus += 1

# Menampilkan hasil
print("Jumlah mahasiswa lulus:", lulus)
print("Jumlah mahasiswa tidak lulus:", tidak_lulus)
```

```python
# Latihan 5
nilai_ujian = [85, 72, 65, 90, 55, 78, 60, 88]

# Fungsi konversi nilai angka menjadi huruf
def konversi_nilai(n):
    if n >= 85:
        return "A"
    elif n >= 70:
        return "B"
    elif n >= 60:
        return "C"
    else:
        return "D"

# Terapkan fungsi ke seluruh nilai menggunakan perulangan
for nilai in nilai_ujian:
    print(nilai, "->", konversi_nilai(nilai))
```
