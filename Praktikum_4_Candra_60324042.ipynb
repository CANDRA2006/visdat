**Menyiapkan Dataset Auto MPG**

```python
import pandas as pd

# Load dataset
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/mpg.csv"
df = pd.read_csv(url)
print(df.shape)

# Mengisi data kosong (missing values) pada kolom horsepower dengan nilai median
df["horsepower"] = df["horsepower"].fillna(df["horsepower"].median())
```

**Line Chart: Tren dari Waktu ke Waktu**

```python
import matplotlib.pyplot as plt
import pandas as pd

# Hitung rata-rata MPG per tahun model
mpg_per_tahun = df.groupby("model_year")["mpg"].mean()

# Buat plot / visualisasi
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(
    mpg_per_tahun.index,
    mpg_per_tahun.values,
    marker="o",
    linestyle="-",
    color="b",
)

ax.set_title("Rata-rata MPG per Tahun Model")
ax.set_xlabel("Tahun Model (19xx)")
ax.set_ylabel("Rata-rata MPG")
ax.grid(True, linestyle="--", alpha=0.6)

plt.tight_layout()
plt.show()
```

**Bar Chart: Membandingkan Kategori**

```python
mpg_per_asal = (df.groupby("origin")["mpg"]
.mean().sort_values(ascending=False))

fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(mpg_per_asal.index,

mpg_per_asal.values, color="steelblue")
ax.set_title("Rata-rata MPG per Asal Mobil")
ax.set_xlabel("Asal Negara")
ax.set_ylabel("Rata-rata MPG")
plt.show()
```

**Scatter Plot: Hubungan Dua Variabel**

```python
import matplotlib.pyplot as plt

# Membuat scatter plot
fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(
    df["horsepower"],
    df["mpg"],
    alpha=0.6,
    color="darkorange",
    edgecolors="k",
)

# Kustomisasi judul dan label
ax.set_title("Tenaga Mesin vs Efisiensi BBM", fontsize=12, pad=10)
ax.set_xlabel("Horsepower (HP)", fontsize=10)
ax.set_ylabel("MPG", fontsize=10)
ax.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()

# Hitung dan cetak nilai korelasi
korelasi = df["horsepower"].corr(df["mpg"])
print(f"Nilai korelasi antara Horsepower dan MPG: {korelasi:.4f}")
```

# Latihan

```python
# Bar Chart horizontal jumlah mobil dengan warna hijau
jumlah = df["origin"].value_counts()

fig, ax = plt.subplots(figsize=(6, 4))
ax.barh(jumlah.index, jumlah.values, color="green")
ax.set_title("Jumlah Mobil per Asal")
ax.set_xlabel("Jumlah Mobil")
plt.tight_layout()
plt.show()

# Line chart rata-rata weight per tahun
weight_per_tahun = df.groupby("model_year")["weight"].mean()

fig2, ax2 = plt.subplots(figsize=(6, 4))
ax2.plot(
    weight_per_tahun.index,
    weight_per_tahun.values,
    marker="o",
    linestyle="-",
    color="b"
)
ax2.set_title("Rata-rata Berat (Weight) per Tahun Model")
ax2.set_xlabel("Tahun Model (19xx)")
ax2.set_ylabel("Rata-rata Berat (lbs)")
ax2.grid(True, linestyle="--", alpha=0.6)

plt.tight_layout()
plt.show()
```
