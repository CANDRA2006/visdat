```python
def hitung_rata2(data):


    total = sum(data)


    jumlah = len(data)


    rata2 = total / jumlah


    return rata2


# Memanggil fungsi


nilai = [80, 90, 75, 85]


hasil = hitung_rata2(nilai)


print("Rata-rata:", hasil)
```

```python
# Latihan: cek kategori nilai mahasiswa


nama = "Candra"


nilai = 100


def kategori_nilai(n):


    if n >= 85:


        return "Sangat Baik"


    elif n >= 70:


        return "Baik"


    else:


        return "Perlu Belajar Lagi"


print(nama, kategori_nilai(nilai))
```
