Impor Libary

```python
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
print("Versi Matplotlib:", matplotlib.__version__)
```

Memuat Dataset Auto MPG

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/mpg.csv"

df = pd.read_csv(url)
print(df.shape)
df.head()
```

Memeriksa dan Menangani Missing Value

```python
print(df.isnull().sum())

# Isi data kosong pada kolom horsepower dengan nilai median
df["horsepower"] = df["horsepower"].fillna(df["horsepower"].median())
print("Data kosong horsepower setelah diisi:", df["horsepower"].isnull().sum())
```

Memahami Figure dan Axes

```python
fig, ax = plt.subplots(figsize=(6, 4))

ax.plot([1, 2, 3, 4], [10, 25, 15, 30])
ax.set_title("Contoh Anatomi Plot")
ax.set_xlabel("Sumbu X")
ax.set_ylabel("Sumbu Y")
```

Membuat Line Chart

```python
mpg_per_tahun = df.groupby("model_year")["mpg"].mean()
print(mpg_per_tahun.round(2).head())

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(mpg_per_tahun.index, mpg_per_tahun.values, marker="o")
ax.set_title("Rata-rata Efisiensi BBM Mobil per Tahun Model")
ax.set_xlabel("Tahun Model (19xx)")
ax.set_ylabel("Rata-rata MPG")
plt.show()
```

Membuat Bar Chart

```python
mpg_per_asal = df.groupby("origin")["mpg"].mean().sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(mpg_per_asal.index, mpg_per_asal.values, color="steelblue")
ax.set_title("Rata-rata Efisiensi BBM Berdasarkan Asal Mobil")
ax.set_xlabel("Asal Negara (origin)")
ax.set_ylabel("Rata-rata MPG")
plt.show()
```

Membuat Scatter Plot

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(df["horsepower"], df["mpg"], alpha=0.6)
ax.set_title("Hubungan Tenaga Mesin dan Efisiensi BBM")
ax.set_xlabel("Horsepower (HP)")
ax.set_ylabel("MPG")

plt.show()
print("Korelasi:", round(df["horsepower"].corr(df["mpg"]), 3))
```

Membandingkan Ketiga Jenis Grafik

## Perbandingan Jenis Grafik
| Grafik | Variabel | Pertanyaan | Keterbatasan |
|--------|----------|------------|--------------|
| Line | model_year, mpg | Bagaimana tren rata-rata efisiensi bahan bakar (MPG) dari tahun ke tahun? | Hanya memperlihatkan nilai agregat rata-rata per tahun, sehingga tidak menggambarkan sebaran atau variasi data MPG pada tahun yang sama. |
| Bar | origin, mpg | Negara asal manakah yang memproduksi mobil dengan rata-rata efisiensi bahan bakar (MPG) tertinggi? | Hanya membandingkan nilai rata-rata antar kategori besar (negara asal) tanpa memperlihatkan distribusi detail maupun jumlah sampel mobil dari masing-masing negara. |
| Scatter | horsepower, mpg | Bagaimana hubungan/korelasi antara tenaga mesin (Horsepower) dengan efisiensi bahan bakar (MPG)? | Meskipun menunjukkan pola korelasi yang jelas, grafik ini tidak dapat secara langsung membuktikan hubungan sebab-akibat (kausalitas) secara mutlak tanpa analisis statistik lebih lanjut. |

Menyimpan Grafik sebagai File Gambar

```python
# Membuat dan menyimpan Line Chart
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(mpg_per_tahun.index, mpg_per_tahun.values, marker="o")
ax.set_title("Rata-rata Efisiensi BBM Mobil per Tahun Model")
ax.set_xlabel("Tahun Model (19xx)")
ax.set_ylabel("Rata-rata MPG")
fig.savefig("line_mpg_tahun.png", dpi=150, bbox_inches="tight")
plt.show()

# Membuat dan menyimpan Bar Chart
fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(mpg_per_asal.index, mpg_per_asal.values, color="steelblue")
ax.set_title("Rata-rata Efisiensi BBM Berdasarkan Asal Mobil")
ax.set_xlabel("Asal Negara (origin)")
ax.set_ylabel("Rata-rata MPG")
fig.savefig("bar_mpg_asal.png", dpi=150, bbox_inches="tight")
plt.show()

# Membuat dan menyimpan Scatter Plot
fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(df["horsepower"], df["mpg"], alpha=0.6)
ax.set_title("Hubungan Tenaga Mesin dan Efisiensi BBM")
ax.set_xlabel("Horsepower (HP)")
ax.set_ylabel("MPG")
fig.savefig("scatter_hp_mpg.png", dpi=150, bbox_inches="tight")
plt.show()

# Melakukan download file yang sudah dibuat
from google.colab import files
files.download("line_mpg_tahun.png")
files.download("bar_mpg_asal.png")
files.download("scatter_hp_mpg.png")
```

Ringkasan Visualisasi Dasar

- Dataset Auto MPG berisi **398** baris dan **9** kolom.

- Grafik yang paling informatif menurut saya adalah **Scatter Plot**, karena **secara langsung memperlihatkan sebaran data secara individual serta tren hubungan korelasi yang jelas dan kuat antara tenaga mesin dengan efisiensi bahan bakar tanpa menyembunyikan variasi datanya**.

# Latihan Mandiri

Latihan 1

```python
jumlah_mobil = df["origin"].value_counts()

fig, ax = plt.subplots(figsize=(6, 4))
ax.barh(jumlah_mobil.index, jumlah_mobil.values, color="seagreen")
ax.set_title("Jumlah Mobil Berdasarkan Asal Negara")
ax.set_xlabel("Jumlah Mobil")
ax.set_ylabel("Asal Negara (origin)")
plt.show()
```

Latihan 2

```python
weight_per_tahun = df.groupby("model_year")["weight"].mean()

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(weight_per_tahun.index, weight_per_tahun.values, linestyle="--", color="red", marker="s")
ax.set_title("Rata-rata Bobot Mobil (Weight) per Tahun Model")
ax.set_xlabel("Tahun Model (19xx)")
ax.set_ylabel("Rata-rata Bobot (lbs)")
plt.show()
```

Latihan 3

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(df["weight"], df["mpg"], alpha=0.5, color="green")
ax.set_title("Hubungan Bobot Mobil (Weight) dan Efisiensi BBM")
ax.set_xlabel("Weight (lbs)")
ax.set_ylabel("MPG")
plt.show()

korelasi = df["weight"].corr(df["mpg"])
print(f"Korelasi: {korelasi:.3f}")
print("Interpretasi: Terdapat hubungan korelasi negatif yang sangat kuat antara bobot mobil dan efisiensi bahan bakar, yang menunjukkan bahwa semakin berat suatu mobil, maka tingkat efisiensi bahan bakarnya (MPG) akan cenderung semakin rendah.")
```

Latihan 4

```python
mpg_per_silinder = df.groupby("cylinders")["mpg"].mean().sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(mpg_per_silinder.index.astype(str), mpg_per_silinder.values, color="teal")
ax.set_title("Rata-rata MPG Berdasarkan Jumlah Silinder")
ax.set_xlabel("Jumlah Silinder (Cylinders)")
ax.set_ylabel("Rata-rata MPG")
plt.show()

silinder_tertinggi = mpg_per_silinder.idxmax()
mpg_tertinggi = mpg_per_silinder.max()
silinder_terendah = mpg_per_silinder.idxmin()
mpg_terendah = mpg_per_silinder.min()

print(f"Jumlah silinder dengan rata-rata MPG tertinggi: {silinder_tertinggi} silinder ({mpg_tertinggi:.2f} MPG)")
print(f"Jumlah silinder dengan rata-rata MPG terendah: {silinder_terendah} silinder ({mpg_terendah:.2f} MPG)")
```

Latihan 5

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(df["displacement"], df["acceleration"], alpha=0.6, color="purple")
ax.set_title("Hubungan Kapasitas Mesin (Displacement) dan Akselerasi")
ax.set_xlabel("Displacement (cu. in.)")
ax.set_ylabel("Acceleration (m/s²)")
plt.show()

print("Alasan pemilihan grafik: Scatter plot adalah jenis grafik terbaik untuk memvisualisasikan persebaran data dan pola hubungan (korelasi) antara dua variabel numerik secara individual.")
print("Interpretasi: Grafik menunjukkan adanya korelasi negatif yang moderat, yang berarti mobil dengan kapasitas mesin (displacement) yang lebih besar cenderung membutuhkan waktu lebih sedikit untuk berakselerasi (nilai akselerasi lebih rendah/lebih cepat).")
```

# Pertanyaan Diskusi

### Jawaban Pertanyaan Diskusi

1. **Tren Tahunan vs Perbandingan Kategori**:
   * **Alasan**: *Line chart* tepat untuk tren tahunan karena garis kontinu memperlihatkan pola naik-turun kronologis secara intuitif. *Bar chart* tepat untuk kategori karena membandingkan nilai mutlak antar kelompok secara mandiri.
   * **Kejanggalan jika ditukar**: Jika ditukar, *bar chart* tahunan akan terasa sangat penuh dan padat oleh batang-batang vertikal, sedangkan *line chart* antar negara akan menyesatkan karena menghubungkan kategori independen seolah-olah memiliki kontinuitas data.

2. **Korelasi Horsepower vs MPG**:
   * **Pola Hubungan**: Terdapat korelasi negatif yang kuat ($r = -0.773$), artinya semakin tinggi tenaga mesin (*horsepower*), efisiensi bahan bakar (*MPG*) cenderung semakin turun.
   * **Kausalitas**: Korelasi kuat saja tidak cukup untuk membuktikan hubungan sebab-akibat secara statistik karena bisa dipengaruhi oleh variabel lain (seperti berat mobil). Namun secara mekanika, tenaga mesin besar memang memerlukan konsumsi BBM yang lebih banyak.

3. **Sumbu Y Mulai dari Nol**:
   * **Alasan**: Sumbu Y pada *bar chart* harus dimulai dari nol agar tinggi batang secara visual mencerminkan perbandingan nilai yang jujur dan akurat.
   * **Dampak jika dipotong (misal mulai 15)**: Perbedaan kecil antar kategori akan terlihat sangat dramatis dan berlebihan, sehingga menyesatkan persepsi pembaca mengenai skala data sebenarnya.
