 2. Pembersihan Data <br>

a) Hitung jumlah baris duplikat. Hapus duplikasi dan simpan pada `df_bersih`. <br>
b) Isi nilai kosong pada wilayah dengan "Tidak diketahui". <br>

c) Isi nilai kosong kolom numerik `kunjungan_akomodasi` dan `rata_rata_pengeluaran_rp` dengan median masing-masing. <br>


```python
#  Menghitung jumlah baris duplikat
jumlah_duplikat = df_uploaded.duplicated().sum()
print(f"Jumlah baris duplikat terdeteksi: {jumlah_duplikat}")

# Menghapus baris duplikat dan menyimpannya di df_bersih
df_bersih = df_uploaded.drop_duplicates().copy()
print(f"Jumlah baris setelah menghapus duplikat: {len(df_bersih)}")
print()

#  Mengisi nilai kosong di kolom wilayah
df_bersih['wilayah'] = df_bersih['wilayah'].fillna('Tidak diketahui')

# Mengisi nilai kosong di kolom numerik dengan median masing-masing
median_akomodasi = df_bersih['kunjungan_akomodasi'].median()
median_pengeluaran = df_bersih['rata_rata_pengeluaran_rp'].median()

df_bersih['kunjungan_akomodasi'] = df_bersih['kunjungan_akomodasi'].fillna(median_akomodasi)
df_bersih['rata_rata_pengeluaran_rp'] = df_bersih['rata_rata_pengeluaran_rp'].fillna(median_pengeluaran)

# Memastikan semua nilai kosong sudah teratasi
print("Jumlah nilai kosong setelah pembersihan:")
print(df_bersih.isnull().sum())
print()
print(f"Dimensi akhir df_bersih: {df_bersih.shape}")
```

**Laporan Pembersihan Data:**

*   **Jumlah Baris Akhir:** Setelah menghapus 1 baris duplikat yang terdeteksi, dataset bersih `df_bersih` kini memiliki total **360 baris**.
*   **Alasan Pemilihan Median:**
    Median dipilih sebagai metode imputasi karena sifatnya yang tangguh (robust) terhadap pencilan (outliers). Pada data pariwisata seperti kunjungan akomodasi dan rata-rata pengeluaran, sering kali terdapat ketimpangan nilai yang sangat ekstrem antara daerah pariwisata utama dengan daerah lainnya. Jika menggunakan rata-rata (mean), nilai tersebut akan sangat dipengaruhi oleh nilai ekstrem tersebut (condong ke atas), sedangkan median (nilai tengah) mencerminkan representasi data yang lebih stabil dan objektif.

3.  Statistik Deskriptif, Groupby, dan Outlier

```python
# Tampilkan describe() untuk kolom numerik
print("Statistik Deskriptif")
display(df_bersih.describe())
print()

# Hitung total perjalanan_wisnus per provinsi dan sebutkan tiga tertinggi
print("Total Perjalanan Wisnus Per Provinsi")
provinsi_travel = df_bersih.groupby('provinsi')['perjalanan_wisnus'].sum().sort_values(ascending=False)
display(provinsi_travel)
print()
print("Tiga Provinsi Teratas:")
for i, (prov, total) in enumerate(provinsi_travel.head(3).items(), 1):
    print(f"{i}. {prov}: {total:,} perjalanan")
print()

# Hitung total perjalanan per bulan untuk seluruh provinsi dan tahun
print("Total Perjalanan Per Bulan")
bulan_travel = df_bersih.groupby('bulan')['perjalanan_wisnus'].sum()
display(bulan_travel)
print()

# Metode IQR pada rata_rata_pengeluaran_rp
q1 = df_bersih['rata_rata_pengeluaran_rp'].quantile(0.25)
q3 = df_bersih['rata_rata_pengeluaran_rp'].quantile(0.75)
iqr = q3 - q1
batas_bawah = q1 - 1.5 * iqr
batas_atas = q3 + 1.5 * iqr

outliers = df_bersih[(df_bersih['rata_rata_pengeluaran_rp'] < batas_bawah) | (df_bersih['rata_rata_pengeluaran_rp'] > batas_atas)]

print("Analisis Pencilan (IQR)")
print(f"Q1          : {q1}")
print(f"Q3          : {q3}")
print(f"IQR         : {iqr}")
print(f"Batas Bawah : {batas_bawah}")
print(f"Batas Atas  : {batas_atas}")
print(f"Jumlah Outlier: {len(outliers)}")
print()

# e) Buat pivot_table total perjalanan dengan baris wilayah dan kolom tahun
print("Pivot Table (Wilayah vs Tahun)")
pivot_perjalanan = df_bersih.pivot_table(
    values='perjalanan_wisnus',
    index='wilayah',
    columns='tahun',
    aggfunc='sum'
)
display(pivot_perjalanan)
```

**Hasil Analisis dan Temuan:**

*   **Tiga Provinsi dengan Perjalanan Tertinggi:** Berdasarkan pengolahan data, tiga provinsi dengan mobilitas perjalanan wisatawan nusantara paling banyak adalah **Bali** (4,463,804), **Jawa Barat** (4,329,650), dan **Jawa Timur** (4,102,832).
*   **Analisis Tren Bulanan:** Bulan dengan tingkat perjalanan tertinggi adalah **Desember** (bulan 12) dengan total akumulasi mencapai lebih dari 4,2 juta perjalanan. Hal ini sangat logis karena Desember merupakan momentum libur akhir tahun, hari raya Natal, dan pergantian tahun baru di mana mobilitas masyarakat untuk berlibur meningkat secara masif.
*   **Hasil Deteksi Outlier:** Berdasarkan metode IQR pada kolom rata-rata pengeluaran, batas atas yang diperoleh adalah Rp582.936,25 sedangkan batas bawahnya Rp88.444,25. Dari perhitungan ini, terdeteksi sebanyak **17 pencilan (outlier)** di mana nilai pengeluarannya berada di luar batas kewajaran umum.
*   **Temuan dari Pivot Table:** Hasil pengamatan dari tabel silang antara wilayah dan tahun, wilayah **Jawa** mencatat volume perjalanan wisatawan domestik yang paling konsisten tinggi di setiap tahunnya. Hal ini disebabkan oleh kemudahan konektivitas infrastruktur darat di pulau Jawa serta jumlah populasi penduduk yang besar, sehingga aktivitas pariwisata lokal antarprovinsi di wilayah tersebut berjalan sangat aktif dibanding wilayah kepulauan lainnya.

4.  Visualisasi

```python
# a) Bar chart total perjalanan per provinsi untuk tahun 2025
df_2025 = df_bersih[df_bersih['tahun'] == 2025]
provinsi_2025 = df_2025.groupby('provinsi')['perjalanan_wisnus'].sum().sort_values(ascending=False)

plt.figure(figsize=(14, 7))
bars = plt.bar(provinsi_2025.index, provinsi_2025.values, color='steelblue', edgecolor='black', alpha=0.9)

# Menambahkan nilai di atas batang dengan format ribuan
for bar in bars:
    yval = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2.0,
        yval + (max(provinsi_2025.values) * 0.01),
        f'{yval:,}',
        ha='center',
        va='bottom',
        fontsize=8,
        rotation=90
    )

plt.title('Total Perjalanan Wisatawan Nusantara Per Provinsi (Tahun 2025)', fontsize=14, fontweight='bold', pad=20)
plt.xlabel('Provinsi', fontsize=12, labelpad=10)
plt.ylabel('Total Perjalanan', fontsize=12, labelpad=10)
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# b) Subplot 1x3 menggunakan Seaborn
fig, axes = plt.subplots(1, 3, figsize=(22, 7))

# (i) Histogram perjalanan_wisnus dengan KDE
sns.histplot(data=df_bersih, x='perjalanan_wisnus', kde=True, ax=axes[0], color='#4c72b0')
axes[0].set_title('Distribusi Perjalanan Wisnus', fontsize=13, fontweight='bold', pad=12)
axes[0].set_xlabel('Jumlah Perjalanan Wisnus', fontsize=11)
axes[0].set_ylabel('Frekuensi', fontsize=11)
axes[0].grid(True, linestyle='--', alpha=0.4)

# (ii) Boxplot rata_rata_pengeluaran_rp per wilayah
sns.boxplot(data=df_bersih, x='wilayah', y='rata_rata_pengeluaran_rp', ax=axes[1], hue='wilayah', palette='muted', legend=False)
axes[1].set_title('Rata-rata Pengeluaran per Wilayah', fontsize=13, fontweight='bold', pad=12)
axes[1].set_xlabel('Wilayah', fontsize=11)
axes[1].set_ylabel('Pengeluaran (Rupiah)', fontsize=11)
# Mengunci posisi sumbu X terlebih dahulu sebelum merotasi labelnya
axes[1].set_xticks(axes[1].get_xticks())
axes[1].set_xticklabels(axes[1].get_xticklabels(), rotation=45, ha='right', fontsize=9)
axes[1].grid(True, linestyle='--', alpha=0.4)

# (iii) Scatter plot perjalanan_wisnus vs kunjungan_akomodasi
sns.scatterplot(
    data=df_bersih,
    x='perjalanan_wisnus',
    y='kunjungan_akomodasi',
    hue='tahun',
    palette='coolwarm',
    alpha=0.8,
    edgecolor='w',
    ax=axes[2]
)
axes[2].set_title('Hubungan Perjalanan & Kunjungan Akomodasi', fontsize=13, fontweight='bold', pad=12)
axes[2].set_xlabel('Perjalanan Wisnus', fontsize=11)
axes[2].set_ylabel('Kunjungan Akomodasi', fontsize=11)
axes[2].grid(True, linestyle='--', alpha=0.4)

plt.tight_layout()

# Menyimpan berkas gambar terlebih dahulu
plt.savefig('uts_pariwisata.png', dpi=300)
plt.show()
```

```python
# c) Heatmap korelasi kolom numerik yang relevan
kolom_numerik = ['perjalanan_wisnus', 'kunjungan_akomodasi', 'kunjungan_objek_wisata', 'rata_rata_pengeluaran_rp']
matriks_korelasi = df_bersih[kolom_numerik].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    matriks_korelasi,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.8,
    linecolor='white',
    vmin=-1,
    vmax=1,
    annot_kws={'size': 11, 'weight': 'bold'}
)

plt.title('Matriks Korelasi Variabel Numerik Pariwisata', fontsize=13, fontweight='bold', pad=15)
plt.tight_layout()
plt.show()
```
